```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]
#line plot
plt.plot(x,y)
plt.title('graph')

plt.xlabel("x axis")
plt.ylabel("y axis")


plt.plot(x,y,marker='*',color='b')
plt.show()
```

```python
#bar chart
course=["ece","eee","cse"]
fees=[80,70,100]
plt.bar(course,fees)
plt.title('college fee')

plt.xlabel("course")
plt.ylabel("fees k")

#plt.show()

plt.plot(course,fees,marker='o',color='g')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

data = np.random.randn(2000)
plt.hist(data, bins=100, color='skyblue', edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Basic Histogram Example')
plt.show()
```

```python
#scatter chart
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 8, 7]

plt.scatter(x, y)

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Basic Scatter Plot")

plt.show()
```

```python
#pie chart
course=["ece","eee","cse"]
fees=[80,70,100]
plt.pie(fees,labels=course,autopct='%1.1f%%')
plt.title('college fee')

plt.xlabel("course")
plt.ylabel("fees k")
plt.show()
```
